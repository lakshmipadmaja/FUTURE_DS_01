```python
# STEP 8.3 - Financial Consistency

print("=" * 50)
print("FINANCIAL CONSISTENCY CHECK")
print("=" * 50)

# Gross sales minus discount
expected_sales = (
    df_clean["gross_sales"]
    - df_clean["discount_amount"]
)

difference = (
    df_clean["net_sales"] - expected_sales
)

print("Difference Summary:")
display(difference.describe().round(2))

print("\nRecords with Difference > 0.01:")
print((difference.abs() > 0.01).sum())
```

```python
# STEP 8.4 - Financial Formula Investigation

import pandas as pd

print("=" * 55)
print("FINANCIAL FORMULA INVESTIGATION")
print("=" * 55)

# Formula 1: Gross Sales - Discount
formula_1 = (
    df_clean["gross_sales"]
    - df_clean["discount_amount"]
)

# Formula 2: Gross Sales - Discount + Tax
formula_2 = (
    df_clean["gross_sales"]
    - df_clean["discount_amount"]
    + df_clean["tax_amount"]
)

# Formula 3: Gross Sales - Discount + Shipping
formula_3 = (
    df_clean["gross_sales"]
    - df_clean["discount_amount"]
    + df_clean["shipping_cost"]
)

# Formula 4: Gross Sales - Discount + Tax + Shipping
formula_4 = (
    df_clean["gross_sales"]
    - df_clean["discount_amount"]
    + df_clean["tax_amount"]
    + df_clean["shipping_cost"]
)

actual_sales = df_clean["net_sales"]

formulas = {
    "Gross - Discount": formula_1,
    "Gross - Discount + Tax": formula_2,
    "Gross - Discount + Shipping": formula_3,
    "Gross - Discount + Tax + Shipping": formula_4
}

results = []

for name, calculated in formulas.items():

    difference = (actual_sales - calculated).abs()

    results.append({
        "Formula": name,
        "Matching Records": (difference <= 0.01).sum(),
        "Average Difference": round(difference.mean(), 2),
        "Maximum Difference": round(difference.max(), 2)
    })

display(pd.DataFrame(results))
```

```python
# STEP 8.5 - Profit Formula Validation

print("=" * 50)
print("PROFIT FORMULA VALIDATION")
print("=" * 50)

# Possible Profit Formula 1
profit_formula_1 = (
    df_clean["net_sales"]
    - df_clean["product_cost"]
)

# Possible Profit Formula 2
profit_formula_2 = (
    df_clean["net_sales"]
    - df_clean["product_cost"]
    - df_clean["shipping_cost"]
)

# Possible Profit Formula 3
profit_formula_3 = (
    df_clean["net_sales"]
    - df_clean["product_cost"]
    - df_clean["shipping_cost"]
    - df_clean["tax_amount"]
)

actual_profit = df_clean["profit"]

profit_formulas = {
    "Net Sales - Product Cost": profit_formula_1,
    "Net Sales - Product Cost - Shipping": profit_formula_2,
    "Net Sales - Product Cost - Shipping - Tax": profit_formula_3
}

profit_results = []

for name, calculated in profit_formulas.items():

    difference = (actual_profit - calculated).abs()

    profit_results.append({
        "Formula": name,
        "Matching Records": (difference <= 0.01).sum(),
        "Average Difference": round(difference.mean(), 2),
        "Maximum Difference": round(difference.max(), 2)
    })

display(pd.DataFrame(profit_results))
```

```python
# STEP 14.11 - Verify Product Row Reconciliation

import pandas as pd

# Keep original product row order
items_check = product_sales_usd.copy()

# Calculate cumulative quantity and revenue per order
items_check["cumulative_quantity"] = (
    items_check.groupby("order_id")["quantity"].cumsum()
)

items_check["cumulative_revenue"] = (
    items_check.groupby("order_id")["net_sales"].cumsum()
)

items_check["cumulative_profit"] = (
    items_check.groupby("order_id")["profit"].cumsum()
)

# Attach order-level financial values
items_check = items_check.merge(
    df_usd[
        ["order_id", "quantity", "net_sales", "profit"]
    ].rename(columns={
        "quantity": "order_quantity",
        "net_sales": "order_revenue",
        "profit": "order_profit"
    }),
    on="order_id",
    how="left",
    validate="many_to_one"
)

# Check whether a prefix of product rows matches
items_check["prefix_matches"] = (
    (items_check["cumulative_quantity"] ==
     items_check["order_quantity"])
    &
    (
        (items_check["cumulative_revenue"] -
         items_check["order_revenue"]).abs() <= 0.01
    )
    &
    (
        (items_check["cumulative_profit"] -
         items_check["order_profit"]).abs() <= 0.01
    )
)

matched_prefix_orders = (
    items_check.loc[
        items_check["prefix_matches"], "order_id"
    ].nunique()
)

print("=" * 55)
print("PRODUCT ROW RECONCILIATION CHECK")
print("=" * 55)

print("Total USD Orders:", df_usd["order_id"].nunique())

print("Orders with Matching Product Prefix:",
      matched_prefix_orders)

print("Orders without Matching Prefix:",
      df_usd["order_id"].nunique() -
      matched_prefix_orders)

# Specifically check mismatched orders
mismatch_ids = set(mismatched_orders["order_id"])

matched_mismatches = (
    items_check.loc[
        items_check["prefix_matches"]
        & items_check["order_id"].isin(mismatch_ids),
        "order_id"
    ].nunique()
)

print("\nMismatched Orders Explained by Prefix:",
      matched_mismatches)

print("Mismatched Orders Still Unexplained:",
      len(mismatch_ids) - matched_mismatches)
```

```python
# STEP 14.13 - Identify Reconciled Product Rows

# Preserve original row sequence
reconciliation_items = items_check.copy()

# Identify the matching endpoint for each order
matching_endpoints = (
    reconciliation_items[
        reconciliation_items["prefix_matches"]
    ]
    .groupby("order_id")
    .cumcount()
)

print("=" * 55)
print("PRODUCT RECONCILIATION PREPARATION")
print("=" * 55)

print(
    "Orders with Matching Prefix:",
    reconciliation_items.loc[
        reconciliation_items["prefix_matches"],
        "order_id"
    ].nunique()
)

print(
    "Orders with Multiple Matching Endpoints:",
    (
        reconciliation_items[
            reconciliation_items["prefix_matches"]
        ]
        .groupby("order_id")
        .size()
        .gt(1)
        .sum()
    )
)
```

```python
# STEP 14.14 - Build Reconciled Product Dataset

# Work with the original row sequence
reconciled_source = product_sales_usd.copy()

# Position of each product row within its order
reconciled_source["_row_position"] = (
    reconciled_source.groupby("order_id").cumcount()
)

# Find matching endpoint for each order
matching_positions = (
    items_check.loc[
        items_check["prefix_matches"],
        ["order_id"]
    ]
    .copy()
)

matching_positions["_row_position"] = (
    items_check.loc[
        items_check["prefix_matches"]
    ]
    .groupby("order_id")
    .cumcount()
)

# Use the first matching cumulative endpoint
items_positions = items_check.copy()

items_positions["_row_position"] = (
    items_positions.groupby("order_id").cumcount()
)

endpoints = (
    items_positions.loc[
        items_positions["prefix_matches"],
        ["order_id", "_row_position"]
    ]
    .groupby("order_id", as_index=False)
    .first()
    .rename(columns={
        "_row_position": "_last_valid_position"
    })
)

reconciled_source = reconciled_source.merge(
    endpoints,
    on="order_id",
    how="left",
    validate="many_to_one"
)

# Retain only the matching prefix
product_sales_usd_reconciled = (
    reconciled_source.loc[
        reconciled_source["_row_position"]
        <= reconciled_source["_last_valid_position"]
    ]
    .drop(
        columns=[
            "_row_position",
            "_last_valid_position"
        ]
    )
    .copy()
)

print("RECONCILED PRODUCT DATASET CREATED!")

print(
    "Original Product Rows:",
    len(product_sales_usd)
)

print(
    "Reconciled Product Rows:",
    len(product_sales_usd_reconciled)
)

print(
    "Excluded Rows:",
    len(product_sales_usd)
    - len(product_sales_usd_reconciled)
)
```

```python
# STEP 14.15 - Final Financial Reconciliation

final_reconciliation = pd.DataFrame({
    "Metric": ["Revenue", "Profit", "Quantity"],

    "Order_Level": [
        df_usd["net_sales"].sum(),
        df_usd["profit"].sum(),
        df_usd["quantity"].sum()
    ],

    "Product_Level": [
        product_sales_usd_reconciled["net_sales"].sum(),
        product_sales_usd_reconciled["profit"].sum(),
        product_sales_usd_reconciled["quantity"].sum()
    ]
})

final_reconciliation["Difference"] = (
    final_reconciliation["Product_Level"]
    - final_reconciliation["Order_Level"]
)

print("=" * 55)
print("FINAL FINANCIAL RECONCILIATION")
print("=" * 55)

display(final_reconciliation.round(2))

# Validate all metrics
assert (
    final_reconciliation["Difference"].abs() < 0.01
).all(), "Reconciliation failed!"

print("\nALL FINANCIAL METRICS RECONCILED SUCCESSFULLY!")
```

```python
# STEP 18.1 - Prepare RFM Dataset

import pandas as pd

rfm = customer_performance[
    [
        "customer_id",
        "Total_Orders",
        "Total_Spending",
        "Last_Purchase"
    ]
].copy()

rfm["Last_Purchase"] = pd.to_datetime(
    rfm["Last_Purchase"]
)

# Use one day after the latest transaction
# as the analysis reference date
reference_date = (
    pd.to_datetime(df_usd["order_date"]).max()
    + pd.Timedelta(days=1)
)

rfm["Recency"] = (
    reference_date - rfm["Last_Purchase"]
).dt.days

rfm["Frequency"] = rfm["Total_Orders"]

rfm["Monetary"] = rfm["Total_Spending"]

print("=" * 55)
print("RFM CUSTOMER ANALYSIS")
print("=" * 55)

print("Reference Date:", reference_date.date())

print("Total Customers:", len(rfm))

display(
    rfm[
        [
            "customer_id",
            "Recency",
            "Frequency",
            "Monetary"
        ]
    ].head(10)
)

print("\nRFM Summary Statistics:")

display(
    rfm[
        ["Recency", "Frequency", "Monetary"]
    ].describe().round(2)
)
```

```python
# STEP 18.2 - Calculate RFM Scores

import pandas as pd

# Recency Score: Lower recency = Higher score
rfm["R_Score"] = pd.qcut(
    rfm["Recency"].rank(method="first"),
    q=5,
    labels=[5, 4, 3, 2, 1]
).astype(int)

# Frequency Score: Higher frequency = Higher score
rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    q=5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

# Monetary Score: Higher spending = Higher score
rfm["M_Score"] = pd.qcut(
    rfm["Monetary"].rank(method="first"),
    q=5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

# Calculate Total RFM Score
rfm["RFM_Score"] = (
    rfm["R_Score"]
    + rfm["F_Score"]
    + rfm["M_Score"]
)

print("=" * 55)
print("RFM SCORING RESULTS")
print("=" * 55)

display(
    rfm[
        [
            "customer_id",
            "R_Score",
            "F_Score",
            "M_Score",
            "RFM_Score"
        ]
    ].head(10)
)

print("\nRFM Score Distribution:")

display(
    rfm["RFM_Score"]
    .value_counts()
    .sort_index()
)

print("\nMinimum RFM Score:", rfm["RFM_Score"].min())
print("Maximum RFM Score:", rfm["RFM_Score"].max())
```

```python
# STEP 18.3 - RFM Customer Segmentation

def assign_rfm_segment(score):

    if score >= 13:
        return "Champions"

    elif score >= 10:
        return "Loyal Customers"

    elif score >= 7:
        return "Potential Loyalists"

    elif score >= 5:
        return "Needs Attention"

    else:
        return "At Risk"


rfm["RFM_Segment"] = (
    rfm["RFM_Score"].apply(assign_rfm_segment)
)

print("=" * 55)
print("RFM CUSTOMER SEGMENTATION")
print("=" * 55)

display(
    rfm[
        [
            "customer_id",
            "Recency",
            "Frequency",
            "Monetary",
            "RFM_Score",
            "RFM_Segment"
        ]
    ].head(10)
)

print("\nCustomer Segment Counts:")

display(
    rfm["RFM_Segment"].value_counts()
)
```

```python
# STEP 18.4 - RFM Segment Performance Analysis

rfm_segment_analysis = (
    rfm.groupby("RFM_Segment")
    .agg(
        Total_Customers=("customer_id", "nunique"),
        Average_Recency=("Recency", "mean"),
        Average_Frequency=("Frequency", "mean"),
        Average_Monetary=("Monetary", "mean"),
        Total_Revenue=("Monetary", "sum")
    )
    .reset_index()
)

# Revenue Contribution
rfm_segment_analysis["Revenue_Share"] = (
    rfm_segment_analysis["Total_Revenue"]
    / rfm_segment_analysis["Total_Revenue"].sum()
    * 100
)

# Customer Contribution
rfm_segment_analysis["Customer_Share"] = (
    rfm_segment_analysis["Total_Customers"]
    / rfm_segment_analysis["Total_Customers"].sum()
    * 100
)

rfm_segment_analysis = (
    rfm_segment_analysis.sort_values(
        "Total_Revenue",
        ascending=False
    )
)

print("=" * 55)
print("RFM SEGMENT PERFORMANCE ANALYSIS")
print("=" * 55)

display(rfm_segment_analysis.round(2))

print(
    "\nTotal Customers:",
    rfm_segment_analysis["Total_Customers"].sum()
)

print(
    "Total Revenue:",
    round(rfm_segment_analysis["Total_Revenue"].sum(), 2)
)
```

```python
# STEP 23.6 - Prepare Monthly Customer Cohort Data

import pandas as pd

print("=" * 60)
print("MONTHLY CUSTOMER COHORT DATA PREPARATION")
print("=" * 60)

# Check whether df_usd exists
if "df_usd" not in globals():
    raise NameError(
        "df_usd is missing. Run the earlier USD data preparation cells first."
    )

# Prepare cohort dataset
cohort_data = df_usd[
    ["order_id", "customer_id", "order_date"]
].copy()

# Convert date column
cohort_data["order_date"] = pd.to_datetime(
    cohort_data["order_date"]
)

# Extract order month
cohort_data["Order_Month"] = (
    cohort_data["order_date"].dt.to_period("M")
)

# Identify first observed purchase month
cohort_data["Cohort_Month"] = (
    cohort_data.groupby("customer_id")["Order_Month"]
    .transform("min")
)

# Calculate cohort index
cohort_data["Cohort_Index"] = (
    (
        cohort_data["Order_Month"].dt.year
        - cohort_data["Cohort_Month"].dt.year
    ) * 12
    +
    (
        cohort_data["Order_Month"].dt.month
        - cohort_data["Cohort_Month"].dt.month
    )
)

print("Total USD Orders:", len(cohort_data))

print(
    "Total Unique Customers:",
    cohort_data["customer_id"].nunique()
)

print(
    "Total Cohort Months:",
    cohort_data["Cohort_Month"].nunique()
)

display(cohort_data.head())

print("\nCOHORT DATA PREPARATION SUCCESSFUL!")
```

```python
# STEP 23.7 - Monthly Cohort Retention Matrix

print("=" * 60)
print("MONTHLY CUSTOMER COHORT RETENTION")
print("=" * 60)

# Count unique active customers in each cohort/month
cohort_counts = (
    cohort_data.groupby(
        ["Cohort_Month", "Cohort_Index"]
    )["customer_id"]
    .nunique()
    .reset_index(name="Active_Customers")
)

# Create cohort matrix
cohort_matrix = cohort_counts.pivot(
    index="Cohort_Month",
    columns="Cohort_Index",
    values="Active_Customers"
)

# Cohort size = customers in first observed month
cohort_sizes = cohort_matrix[0]

# Calculate retention percentages
retention_matrix = (
    cohort_matrix.div(cohort_sizes, axis=0) * 100
)

print("\nCustomer Cohort Sizes:")

display(
    cohort_sizes.rename("New_Customers").to_frame()
)

print("\nMonthly Retention Matrix (%):")

display(retention_matrix.round(2))

print("\nTotal Cohort Customers:", int(cohort_sizes.sum()))

assert int(cohort_sizes.sum()) == 14680

print("\nCOHORT RETENTION MATRIX CREATED SUCCESSFULLY!")
```

```python
# STEP 23.7 - Monthly Cohort Retention Matrix

print("=" * 60)
print("MONTHLY CUSTOMER COHORT RETENTION")
print("=" * 60)

# Count unique active customers in each cohort/month
cohort_counts = (
    cohort_data.groupby(
        ["Cohort_Month", "Cohort_Index"]
    )["customer_id"]
    .nunique()
    .reset_index(name="Active_Customers")
)

# Create cohort matrix
cohort_matrix = cohort_counts.pivot(
    index="Cohort_Month",
    columns="Cohort_Index",
    values="Active_Customers"
)

# Cohort size = customers in first observed month
cohort_sizes = cohort_matrix[0]

# Calculate retention percentages
retention_matrix = (
    cohort_matrix.div(cohort_sizes, axis=0) * 100
)

print("\nCustomer Cohort Sizes:")

display(
    cohort_sizes.rename("New_Customers").to_frame()
)

print("\nMonthly Retention Matrix (%):")

display(retention_matrix.round(2))

print("\nTotal Cohort Customers:", int(cohort_sizes.sum()))

assert int(cohort_sizes.sum()) == 14680

print("\nCOHORT RETENTION MATRIX CREATED SUCCESSFULLY!")
```

```python
# STEP 23.8 - Monthly Cohort Retention Summary

print("=" * 60)
print("CUSTOMER COHORT RETENTION SUMMARY")
print("=" * 60)

# Last observed month in the dataset
last_month = cohort_data["Order_Month"].max()

# Calculate how many months each cohort could be observed
cohort_age = pd.Series(
    [
        (last_month.year - month.year) * 12
        + (last_month.month - month.month)
        for month in retention_matrix.index
    ],
    index=retention_matrix.index
)

retention_summary = []

for month_index in [1, 2, 3, 6, 12]:

    # Include only cohorts old enough for this month
    eligible_cohorts = cohort_age[
        cohort_age >= month_index
    ].index

    eligible_customers = cohort_sizes.loc[
        eligible_cohorts
    ].sum()

    if month_index in cohort_matrix.columns:

        retained_customers = (
            cohort_matrix.reindex(
                index=eligible_cohorts,
                columns=[month_index]
            )[month_index]
            .fillna(0)
            .sum()
        )

    else:
        retained_customers = 0

    retention_rate = (
        retained_customers / eligible_customers * 100
        if eligible_customers > 0
        else np.nan
    )

    retention_summary.append({
        "Month": month_index,
        "Eligible_Customers": int(eligible_customers),
        "Retained_Customers": int(retained_customers),
        "Retention_Rate": retention_rate
    })

retention_summary = pd.DataFrame(retention_summary)

display(retention_summary.round(2))
```

```python
# STEP 25.4 - Prepare Dashboard Data

print("=" * 60)
print("PREPARING FINAL DASHBOARD DATA")
print("=" * 60)

# Use validated USD orders
dashboard_data = df_usd.copy()

dashboard_data["order_date"] = pd.to_datetime(
    dashboard_data["order_date"]
)

# Monthly sales
dashboard_data["Year_Month"] = (
    dashboard_data["order_date"]
    .dt.to_period("M")
    .astype(str)
)

monthly_dashboard = (
    dashboard_data
    .groupby("Year_Month")
    .agg(
        Revenue=("net_sales", "sum"),
        Profit=("profit", "sum"),
        Orders=("order_id", "nunique")
    )
    .reset_index()
    .sort_values("Year_Month")
)

# Regional sales
regional_dashboard = (
    dashboard_data
    .groupby("region")
    .agg(
        Revenue=("net_sales", "sum"),
        Profit=("profit", "sum")
    )
    .reset_index()
    .sort_values("Revenue", ascending=False)
)

# Marketing performance
marketing_dashboard = (
    dashboard_data
    .groupby("marketing_channel", dropna=False)
    .agg(
        Revenue=("net_sales", "sum"),
        Profit=("profit", "sum")
    )
    .reset_index()
    .sort_values("Revenue", ascending=False)
)

# Customer loyalty
loyalty_dashboard = (
    customer_loyalty
    .groupby("Loyalty_Type")
    .agg(
        Customers=("customer_id", "nunique")
    )
    .reset_index()
)

# Product category data from reconciled product dataset
product_category_column = next(
    (
        col for col in product_sales_usd_reconciled.columns
        if col.lower() in ["category", "product_category"]
    ),
    None
)

product_revenue_column = next(
    (
        col for col in product_sales_usd_reconciled.columns
        if col.lower() in [
            "product_revenue",
            "item_revenue",
            "net_sales",
            "revenue"
        ]
    ),
    None
)

if product_category_column is None or product_revenue_column is None:
    raise ValueError(
        "Product category or revenue column not identified. "
        "Check product_sales_usd_reconciled.columns."
    )

category_dashboard = (
    product_sales_usd_reconciled
    .groupby(product_category_column)
    .agg(
        Revenue=(product_revenue_column, "sum")
    )
    .reset_index()
    .rename(columns={product_category_column: "Category"})
    .sort_values("Revenue", ascending=False)
    .head(10)
)

print("\nMonthly Sales Data:")
display(monthly_dashboard.head())

print("\nRegional Data:")
display(regional_dashboard)

print("\nMarketing Data:")
display(marketing_dashboard.head())

print("\nCustomer Loyalty Data:")
display(loyalty_dashboard)

print("\nCategory Data:")
display(category_dashboard)

print("\nDASHBOARD DATA PREPARATION COMPLETED!")
```

```python
# STEP 25.5 - Dashboard KPI Cards

total_revenue = dashboard_data["net_sales"].sum()
total_profit = dashboard_data["profit"].sum()

total_orders = dashboard_data["order_id"].nunique()
total_customers = dashboard_data["customer_id"].nunique()

profit_margin = (
    total_profit / total_revenue * 100
)

dashboard_kpis = {
    "Revenue": total_revenue,
    "Profit": total_profit,
    "Orders": total_orders,
    "Customers": total_customers,
    "Profit Margin": profit_margin
}

print("=" * 60)
print("DASHBOARD KPI SUMMARY")
print("=" * 60)

for metric, value in dashboard_kpis.items():
    print(f"{metric}: {value:,.2f}")
```

```python
# STEP 25.6 - Professional Interactive Sales Dashboard

fig = make_subplots(
    rows=4,
    cols=2,

    specs=[
        [{"type": "indicator"}, {"type": "indicator"}],
        [{"type": "indicator"}, {"type": "indicator"}],
        [{"type": "xy"}, {"type": "xy"}],
        [{"type": "xy"}, {"type": "xy"}]
    ],

    subplot_titles=[
        "Total Revenue",
        "Total Profit",
        "Total Orders",
        "Total Customers",
        "Monthly Revenue & Profit",
        "Regional Revenue",
        "Top Product Categories",
        "Marketing Channel Revenue"
    ],

    vertical_spacing=0.10,
    horizontal_spacing=0.12
)

# KPI 1 - Revenue
fig.add_trace(
    go.Indicator(
        mode="number",
        value=total_revenue,
        number={
            "prefix": "$",
            "valueformat": ",.0f"
        }
    ),
    row=1,
    col=1
)

# KPI 2 - Profit
fig.add_trace(
    go.Indicator(
        mode="number",
        value=total_profit,
        number={
            "prefix": "$",
            "valueformat": ",.0f"
        }
    ),
    row=1,
    col=2
)

# KPI 3 - Orders
fig.add_trace(
    go.Indicator(
        mode="number",
        value=total_orders,
        number={"valueformat": ",.0f"}
    ),
    row=2,
    col=1
)

# KPI 4 - Customers
fig.add_trace(
    go.Indicator(
        mode="number",
        value=total_customers,
        number={"valueformat": ",.0f"}
    ),
    row=2,
    col=2
)

# Monthly Revenue
fig.add_trace(
    go.Scatter(
        x=monthly_dashboard["Year_Month"],
        y=monthly_dashboard["Revenue"],
        mode="lines+markers",
        name="Monthly Revenue",
        line={"color": "#2563EB"}
    ),
    row=3,
    col=1
)

# Monthly Profit
fig.add_trace(
    go.Scatter(
        x=monthly_dashboard["Year_Month"],
        y=monthly_dashboard["Profit"],
        mode="lines+markers",
        name="Monthly Profit",
        line={"color": "#10B981"}
    ),
    row=3,
    col=1
)

# Regional Revenue
fig.add_trace(
    go.Bar(
        x=regional_dashboard["region"],
        y=regional_dashboard["Revenue"],
        name="Regional Revenue",
        marker_color="#8B5CF6"
    ),
    row=3,
    col=2
)

# Category Revenue
fig.add_trace(
    go.Bar(
        x=category_dashboard["Category"],
        y=category_dashboard["Revenue"],
        name="Category Revenue",
        marker_color="#F59E0B"
    ),
    row=4,
    col=1
)

# Marketing Revenue
fig.add_trace(
    go.Bar(
        x=marketing_dashboard["marketing_channel"],
        y=marketing_dashboard["Revenue"],
        name="Marketing Revenue",
        marker_color="#EC4899"
    ),
    row=4,
    col=2
)

fig.update_layout(
    title={
        "text": (
            "BUSINESS SALES ANALYTICS DASHBOARD"
            "<br><sup>USD | Completed & Paid Orders</sup>"
        ),
        "x": 0.5,
        "font": {"size": 24}
    },

    template="plotly_white",

    height=1500,
    width=1400,

    showlegend=True,

    paper_bgcolor="#F8FAFC",

    font={
        "family": "Arial",
        "size": 12
    }
)

fig.update_xaxes(
    tickangle=-35,
    row=4
)

fig.show()

print("INTERACTIVE BUSINESS DASHBOARD CREATED!")
```
